```python
# Import packages
from pathlib import Path
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
# Import data
df = pd.read_csv(Path('data/processed') / 'panel_data.csv')
```

```python
# Take a look at data
df.head()
```

# Run regression

## Variable descriptions
- `tariff_lag1`: Import-weighted tariff rate (t-1)
- `hc`: Human capital index
- `csh_g`: Share of government consumption at current PPPs
- `pl_i`: Price level of capital formation
- `csh_i`: Share of gross capital formation at current PPPs
- `pop_growth`: Log population growth rate
- `income_group`: World Bank income classification dummies

```python
formula = """
log_gdppc_growth
~ tariff_lag1
+ hc + csh_g + pl_i + csh_i + pop_growth
+ C(income_group)
"""

variables = ['log_gdppc_growth','tariff_lag1',
             'hc','csh_g','pl_i','csh_i','pop_growth','income_group','country']
df_model = df[variables].dropna()

model = smf.ols(formula, data=df_model)
results = model.fit(cov_type="cluster", cov_kwds={"groups": df_model["country"]})

print(results.summary())
```
